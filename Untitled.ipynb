```python
import pandas as pd
#averge net engagent by hours of days 
#ensure post_time is in datatime format
df['post_time']=pd.to_datetime(df['post_time'], errors='coerce')
df['post_hour']=df['post_time'].dt.hour
df['net_engagement']=df['likes']+df['comments']+df['shares']
#group by hour
hourly_engagement=df.groupby('post_hour')['net_engagement'].mean().reset_index()
#line plot
plt.figure(figsize=(10,3))
sns.lineplot(data=hourly_engagement, x='post_hour',y='net_engagement',marker='o',color='blue')
plt.xlabel('hours of day')
plt.ylabel('average net engagement')
plt.title('average net engagement by hours day')
plt.grid(True,ls='--',alpha=0.5)
plt.tight_layout()
plt.plot()
plt.show()
              
```
